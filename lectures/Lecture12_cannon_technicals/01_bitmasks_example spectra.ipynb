```python
import numpy as np
from matplotlib import pyplot as plt
from astropy.io import fits
from astropy.table import Table
from astropy import units as u 
import requests
import io
import os
from joblib import Memory

plt.rc('font', family='serif', size=14)

# Cache downloaded FITS files to avoid repeated downloads
cache_dir = os.path.join(os.path.dirname(os.path.abspath(".")), ".fits_cache")
memory = Memory(cache_dir, verbose=0)
```

Spectra and all associated data are stored in .fits files.
- Great algorithms for compression
- Can hold many different data types
- Contain headers, images, **spectra**, tables, etc.

APOGEE Spectra are all hosted on the `data.sdss.org` website so we can use Python to directly download spectra from online! `Astropy` has functionality directly built into the fits module to do it for you. We just have to find the spectra/stars we care about for our problem. Eg. https://data.sdss.org/sas/dr17/apogee/spectro/redux/dr17/stars/apo25m/M15/

```python
base = 'https://data.sdss.org/sas/dr17/apogee/spectro/redux/dr17/stars/apo25m/'
field = 'M15'
filename = 'apStar-dr17-2M21235315+1244123.fits'

@memory.cache
def download_fits(url, timeout=120):
    """Download a FITS file from a URL with a configurable timeout, caching the result."""
    resp = requests.get(url, timeout=timeout)
    resp.raise_for_status()
    return resp.content

try:
    url = base + f"{field}/{filename}"
    fits_bytes = download_fits(url)
    with fits.open(io.BytesIO(fits_bytes)) as hdul:
        header = hdul[0].header
        spectra = hdul[1].data
        print('Shape of Spectra', spectra.shape)
        errors = hdul[2].data
        print('Shape of Errors', errors.shape)
        bitmasks = hdul[3].data
        print('Shape of Bitmasks', bitmasks.shape)

except Exception as err:
    print(f"Error downloading {filename}: {err}")
```

We have 4 different spectra for this star.
- **Spectra 1: The Co-added Spectra of all the visits**
- Spectra 2: Co-added in a different way
- Spectra 3: Visit 1
- Spectra 4: Visit 2

We now have the fluxes (spectra), errors, and a bitmask for the "bad" pixels, but we need a wavelength grid! Let's take a look at the header for that.

```python
spectrum = spectra[0]
error = errors[0]
bitmask = bitmasks[0]

header
```

The important entries are CRVAL1, CDELT1, and NWAVE. We can use these 3 values to build the wavelength grid. The grid is 'LOG-LINEAR' as mentioned in entry CTYPE1. So,

$$ \lambda_i = 10^{{\rm CRVAL1}+i\cdot{\rm CDELT1}} $$

```python
wavelength = 10.0 ** np.arange(header['CRVAL1'], header['CRVAL1']+header['NWAVE']*header['CDELT1'], header['CDELT1']) * u.angstrom
```

Now we can plot our spectra!

```python
plt.rc('text', usetex=False)

fig, axes = plt.subplots(3, 1, sharex=True, layout='constrained', figsize=(15, 10))
xlabels = (r'Flux [$10^{-17}$ erg/s/cm$^2$/Å]', r'Flux Error [$10^{-17}$ erg/s/cm$^2$/Å]', 'Bitmask')

for ax, data, xlabel in zip(axes, (spectrum, error, bitmask), xlabels):
    ax.plot(wavelength, data, c='k')
    ax.set_xlim(wavelength.min().value, wavelength.max().value)
    ax.set_ylabel(xlabel, fontsize=14)
    ax.grid(True)
fig.supxlabel(r'Wavelength [Å]', fontsize=14)
plt.show()
```

What is going on with the bitmask? That does not look like a boolean mask! Bitmasks are a somewhat outdated, but widely used way to have multiple boolean flags all stored in 1 number. The bitmask only seems to take on certain integer values. Any integer can be represented in base 2 also known as binary. For example, 9 is 1001 because 9 = 8x1 + 4x0 + 2x0 + 1x1. Each of these bits represent a different flag. If we take a look at the website the 0th bit and the 3rd bit being set means that "Pixel marked as BAD in bad pixel mask or from strong persistence jump" and "Pixel marked as unfixable in ap3d". **In this lab we only care if bits 0-7 and 12 are marked.**

```python
target_bits = [0,1,2,3,4,5,6,7,12] # the bits we care about
target_number = np.sum(2**np.array(target_bits)) # turning our target bits into an integer
boolean_mask = np.zeros_like(bitmask, dtype=bool) # initializing an empty boolean mask
boolean_mask[(bitmask & target_number) != 0] = True # doing a bitwise AND with the bitmask from the spectra and our target integer

# keep original errors for plotting before overwriting
error_original = errors[0].copy()
error[boolean_mask] = 1e6

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, layout='constrained', figsize=(10, 7),
                                height_ratios=[3, 1])
ax1.plot(wavelength[~boolean_mask], spectrum[~boolean_mask], c='k')
ax1.set_xlim(wavelength.min().value, wavelength.max().value)
ax1.set_ylabel(r'Flux [$10^{-17}$ erg/s/cm$^2$/\AA]', fontsize=14)
ax1.grid(True)

ax2.plot(wavelength[~boolean_mask], error_original[~boolean_mask], c='k')
ax2.set_ylim(0,30)
ax2.set_ylabel(r'Flux Error', fontsize=14)
ax2.grid(True)

# shade regions where error > 20 or missing (masked out)
bad_error = np.ones(len(wavelength), dtype=bool)
bad_error[~boolean_mask] = error_original[~boolean_mask] > 20

# find contiguous bad regions and shade them
wv = wavelength.value
diff = np.diff(bad_error.astype(int))
starts = np.where(diff == 1)[0] + 1
ends = np.where(diff == -1)[0] + 1
if bad_error[0]:
    starts = np.concatenate(([0], starts))
if bad_error[-1]:
    ends = np.concatenate((ends, [len(bad_error)]))

for s, e in zip(starts, ends):
    for ax in (ax1, ax2):
        ax.axvspan(wv[s], wv[min(e, len(wv)-1)], color='grey', alpha=0.15)

fig.supxlabel(r'Wavelength [\AA]', fontsize=14)
plt.show()
```
